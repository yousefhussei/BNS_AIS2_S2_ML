```python
def Calculator():
    print("--- Welcome to the Calculator! ---")
    while True:
        print("\nSelect an operation:")
        print("1. Addition (+)")
        print("2. Subtraction (-)")
        print("3. Multiplication (*)")
        print("4. Division (/)")

        choice = input("Enter your choice...")

        if choice == "5":
            print("thank you, bye bye..!")
            break

        num1 = float(input("Enter frist number: "))
        num2 = float(input("Enter second number: "))

        if choice == "1":
            print(f"The result of adding {num1} and {num2} is {num1 + num2}")

        elif choice == "2":
            print(f"The result of subtracting {num1} from {num2} is {num1 - num2}")

        elif choice == "3":
            print(f"The result of multiplying {num1} from {num2} is {num1 * num2}")

        elif choice == "4":
            if num2 == 0:
                print("Cannot divide by zero!")
            else:
                print(f"The result of dividing {num1} from {num2} is {num1 / num2}")

        else:
            print("Invalid choice!")



```

```python
Calculator()
```
