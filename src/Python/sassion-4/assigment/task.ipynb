```python
##  1

def  multiplication_table(number:int)-> None:
    ''' 
    Print the multiplication table of a number
     
    args: 
        num(int): user input the int number 
    
    return:
        None: print the multiplication table 
        
    '''

    for i in range(1,11):
        print(f"{number} x {i} = {number * i}")



number = int(input("Enter a number: "))
multiplication_table(number)

```

```python
##  2

def is_prime(num: int) -> bool:
    '''
    Check if a number is prime

    args:
        num(int): user input the int number

    return:
        bool: return True if the number is prime
    '''

    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


def twin_primes() -> None:
    '''
    Print twin prime numbers less than 1000

    return:
        None: print the twin prime numbers
    '''

    for num in range(3, 1000, 2):
        if is_prime(num) and is_prime(num + 2):
            print(num, "->", num + 2)


twin_primes()
```

```python
##  3


def prime_factors(num: int) -> None:
    '''
    Calculate the prime factors of a number

    args:
        num(int): user input the int number

    return:
        None: print the prime factors
    '''

    i = 2

    while num > 1:
        if num % i == 0:
            print(i)
            num = num // i
        else:
            i += 1


number = int(input("Enter a number: "))
prime_factors(number)


    
```

```python
##  4


def to_binary(num: int) -> str:
    '''
    Convert decimal number to binary

    args:
        num(int): user input the int number

    return:
        str: return the binary number
    '''

    binary = ""

    while num > 0:
        remainder = num % 2
        binary = str(remainder) + binary
        num = num // 2

    return binary


number = int(input("Enter a number: "))
print(to_binary(number))

```

```python
##  5

def perfect_numbers(start: int, end: int) -> None:

    for num in range(start, end + 1):

        total = 0

        for i in range(1, num // 2 + 1):
            if num % i == 0:
                total += i

        if total == num:
            print(num)


perfect_numbers(0, 100)
```
