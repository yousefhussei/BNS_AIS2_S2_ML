```python
def func(item,lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst

print(func(1))
print(func(2))
print(func(3))
```

```python
def train_model(model_name,*datasets,**hyperparameters):
    print(f"model_name: {model_name}")

    print("\nDatasets:")
    for dataset in datasets:
        print(f"- {dataset}")

    print("\nHyperparameters:")
    for key, value in hyperparameters.items():
        print(f"- {key}: {value}")

    
```

```python
train_model(
    "Fraud Detection",
    "train.csv",
    "validation.csv",
    "test.csv",
    learing_rate = 0.001,
    epochs = 100,
    baths_zise = 64,
    optimizer = "Adam",
    early_stopping = True

)
```
