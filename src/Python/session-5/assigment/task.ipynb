```python
## Task 1 — Factorial

class Factorial:
    '''
    A class used to calculate the factorial of a number using recursion.

    Attributes:
        num (int): The number whose factorial will be calculated.
    '''
    def __init__(self,num:int):
        """
        Initialize the Factorial object.

        Args:
            num (int): The input number.
        """
        self.num = num

    def calculate(self)-> int:
        ''' 
        Calculate n! useing recursion


        return:
            num(int):return the factorial

        '''
        if self.num == 0:
            return 1
        
        return self.num * Factorial(self.num-1).calculate()

task1 = Factorial(5)

factorial1 = task1.calculate()

print("Factorial: ", factorial1)
    
        
```

```python
##  Task 2 — Check Prime Number
class PrimeNumper:
    '''
    A class used to check a number is prime or not.

    Attributes:
        num (int): The number to be checked.
    '''
    def __init__(self,num:int):
        """
        Initialize the PrimeNumper object.

        Args:
            num (int): The input number.
        """
        self.num = num

    def is_prime(self)-> bool:
        '''
        Check for the number is prime or not

        return:
            bool: True if the number is prime, False otherwise
        '''
        if self.num < 2:
            return False

        for i in range(2, self.num):
            if self.num % i == 0:
                return False

        return True

    
task2 = PrimeNumper(17)

prime_numper = task2.is_prime()

print("is prime: ", prime_numper)
```

```python
class CommonDivisors:
    '''
    A class used to calculate the common divisors of two numbers.

    Attributes:
        num1 (int): The first number.
        num2 (int): The second number.
    '''

    def __init__(self, num1: int, num2: int):
        '''
        Initialize the CommonDivisors object.

        Args:
            num1 (int): The first number.
            num2 (int): The second number.
        '''
        self.num1 = num1
        self.num2 = num2

    def common_divisor(self)-> list[int]:

        '''
        this function help to calc the common divisor



        return:
            list[int]: return the common divisor of two numbers

        '''
        limit = min(self.num1,self.num2)
        divisors = []

        for divisor in range(1, limit+1):
            if self.num1 % divisor == 0 and self.num2 % divisor == 0:
                divisors.append(divisor)
        return divisors


task3 = CommonDivisors(10,20)

common_divisors = task3.common_divisor()

print("common divisor: ",common_divisors)
          
```
