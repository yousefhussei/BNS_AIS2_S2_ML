```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    pass



a = Bird()
a.move()
```

```python
class Rectangel:
    _length = 5 
    width = 10

    def area(self):
        return self._length * self.width

r = Rectangel()
print(r._length)
print(r.width)
print(r.area())
```

```python
class Rectangel:
    __length = 5 
    width = 10

    def area(self):
        return self.__length * self.width
r = Rectangel()
print(r._Rectangel__length)
print(r.width)
print(r.area())
print(r.__dict__)
```

```python
class Parent:
    def __init__(self)-> None:
        self.__value = 10

class Child(Parent):
    def __init__(self)-> None:
        super().__init__()
        self.__value = 20
```

```python
c = Child()
print(c.__dict__)
```

```python
class shape:
    def calcolate_area(self):
        pass

class Circle(shape):
    def __init__(self,redius)->None:
        self.redius = redius

    def calcolate_area(self):
        return 3.14 * (self.redius)

class square(shape):
    def __init__(self,side)-> None:
        self.side = side

    def calcolate_area(self):
        return self.side * self.side

x = Circle(5)
y = square(4)
print("Area: ", x.calcolate_area())
print("Area: ", y.calcolate_area())
```
