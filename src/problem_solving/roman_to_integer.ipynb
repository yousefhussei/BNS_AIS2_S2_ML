```python
class Solution:
    def romanToInt(self, s: str) -> int:
        roman = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }


        total = 0

        for i in range(len(s)):
            # بنطرح لما يبقا الرقم الاول اصغر من الرقم التاني
            if i + 1 < len(s) and roman[s[i]] < roman[s[i + 1]]:
                total -= roman[s[i]]
            else:
                # بنجمع لما يبقا الرقم الاول اكبر من الرقم التاني
                total += roman[s[i]]

        return total

print(Solution().romanToInt("MCMXCIV"))
print(Solution().romanToInt("IV"))
```
