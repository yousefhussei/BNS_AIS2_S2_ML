# <a id="1"> 1 ) . What is the distribution for User type?</a>

```python
## I will import seaborn library to support me in making visualizations 

import seaborn as sns
plt.figure(figsize = (10,5), dpi = 100)

plot = sns.countplot(data=df, x='user_type')

plt.xlabel('User distribution')
plt.ylabel('Frequency')
plt.title("Represent the user's distribution depending on being a part of the Trips Program")

plt.show()
```

The vast majority of trips were made by **Subscribers** (annual/monthly members) rather than one-off **Customers**. Subscribers account for roughly 90% of all trips in this dataset, which suggests the service is mainly used by regular, repeat commuters rather than tourists or occasional riders.

# <a id="2"> 2 ) What is the distribution for the bike share users on all Trips?</a>

```python
plt.figure(figsize = (8,5), dpi = 100)

sb.countplot(data=df, x='bike_share_for_all_trip')

plt.xlabel('Part of Bike Share For All Program')
plt.ylabel('Number of Trips')
plt.title("Distribution of Trips by Bike Share For All Program")

plt.show()
```

# <a id="3"> 3 ) What is the distribution for users' ages? </a>

```python
plt.figure(figsize = (8,5), dpi = 100)

bins = np.arange(df['age'].min(), df['age'].max()+2, 2)
plt.hist(data=df, x='age', bins=bins)

plt.xlabel('Age (years)')
plt.ylabel('Number of Trips')
plt.title("Distribution of Users' Ages")

plt.show()
```

> ### Analysis and exploration

The age distribution is unimodal and right-skewed, peaking in the late twenties to mid-thirties. Most riders are between 25 and 40 years old, and the number of trips drops off steadily for older age groups.

```python
plt.figure(figsize = (8,5), dpi = 100)

sb.boxplot(data=df, x='age')
plt.xlabel('Age (years)')
plt.title("Boxplot of Users' Ages (checking for outliers)")

plt.show()

print('Number of users older than 80:', (df['age'] > 80).sum())
print('Max age in dataset:', df['age'].max())
```

# <a id="5">  5 ) . What is the distribution for the trip duration in minutes? </a>

```python
## Review the statistic for trip duration in minutes

df['duration_min'].describe()
```

```python
# Make visualization to easy get insight
plt.figure(figsize = (8,5), dpi = 100)

# most trips are short, so we clip the view to under 60 minutes for readability
bins = np.arange(0, 61, 2)
plt.hist(data=df[df['duration_min'] < 60], x='duration_min', bins=bins)

plt.xlabel('Trip Duration (minutes)')
plt.ylabel('Number of Trips')
plt.title("Distribution of Trip Duration in Minutes (trips under 60 min)")

plt.show()
```

# <a id="8"> 8) . What the average duration of trip for the categories of user type?</a>

```python
# Distribution of Age by user types
plt.figure(figsize = (8,5), dpi = 100)

sb.boxplot(data=df[df['duration_min'] < 60], x='user_type', y='duration_min')

plt.xlabel('User Type')
plt.ylabel('Trip Duration (minutes)')
plt.title("Trip Duration by User Type (trips under 60 min)")

plt.show()

df.groupby('user_type', observed=True)['duration_min'].mean()
```

> ### Analysis and exploration

**Customers** (non-members using single/day passes) take noticeably longer trips on average than **Subscribers**. This makes sense: subscribers tend to use the bikes for quick, routine commutes, while customers may be tourists or occasional users taking more leisurely rides.

# <a id="9">    9 ) . How long is the duration of the trip for each gender in minutes?  </a>

```python
plt.figure(figsize = (8,5), dpi = 100)

sb.boxplot(data=df[df['duration_min'] < 60], x='member_gender', y='duration_min')

plt.xlabel('Gender')
plt.ylabel('Trip Duration (minutes)')
plt.title("Trip Duration by Gender (trips under 60 min)")

plt.show()

df.groupby('member_gender', observed=True)['duration_min'].mean()
```

> ### Analysis and exploration

Trip duration is fairly similar across genders, though members identifying as **Other** and **Female** tend to have slightly longer median trip durations than **Male** members.

# <a id="10 "> 10 ) . How long is the duration of the trip for each user_type in minutes?  </a>

```python
plt.figure(figsize = (8,5), dpi = 100)

sb.pointplot(data=df, x='user_type', y='duration_min',
             linestyle='none', errorbar=('ci', 95))

plt.xlabel('User Type')
plt.ylabel('Mean Trip Duration (minutes)')
plt.title("Mean Trip Duration by User Type (with 95% CI)")

plt.show()
```

> ### Analysis and exploration

This point plot confirms the boxplot finding from question 8: **Customers** have a noticeably higher mean trip duration than **Subscribers**, and the tight confidence intervals (due to the large sample size) confirm this difference is not just noise.

## Limitations

- We had missing data in `member_birth_year`, `member_gender`, and the station id/name columns. We addressed this by dropping the affected rows (about 4.6% of the data in total), which is straightforward but does slightly reduce the sample size and could introduce bias if the missingness wasn't random.

- The `start_time`/`end_time` columns in this extract only contained a truncated `MM:SS.s` value rather than a full date-time, so we were unable to explore time-of-day, day-of-week, or seasonal usage patterns, which would otherwise have been a natural and valuable part of this analysis.

- Age outliers (some riders recorded as over 100 years old) suggest some `member_birth_year` values are unreliable data-entry errors; a stricter cleaning step (e.g. capping age at a plausible maximum) could improve future analysis.
