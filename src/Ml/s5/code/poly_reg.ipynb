```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import r2_score, mean_squared_error 
```

```python
data = pd.read_csv("Salary_Data.csv")
data.head()
```

```python
x = data.iloc[:,:-1]
y = data.iloc[:,1]
x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.8, random_state=60)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x_train, y_train)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=5)
x_poly_trian = poly.fit_transform(x_train)
x_poly_test = poly.transform(x_test)
```

```python
poly_model = LinearRegression()
poly_model.fit(x_poly_trian, y_train )

y_poly_pred_train = poly_model.predict(x_poly_trian)
y_poly_pred_test = poly_model.predict(x_poly_test)
```

```python
x_range = np.linspace(x.min(), x.max(), 100).reshape(-1,1)
x_range_poly =poly.transform(x_range) 
y_range_red = poly_model.predict(x_range_poly)
```

```python
plt.scatter(x_train, y_train, color ="red")
plt.scatter(x_test, y_test, color ="green")
plt.plot(x_train, lin_reg.predict(x_train), color="blue")
```

```python
y_lin_pred_train = lin_reg.predict(x_train)
y_lin_pred_test = lin_reg.predict(x_test)
train_lin_r2 = r2_score(y_train, y_lin_pred_train)
test_lin_r2 = r2_score(y_test, y_lin_pred_test)
print("train_lin_r2", train_lin_r2)
print("test_lin_r2", test_lin_r2)
```

```python
trian_poly_r2 = r2_score(y_train,y_poly_pred_train)
test_poly_r2 = r2_score(y_test,y_poly_pred_test)
print("trian_poly r2", trian_poly_r2)
print("test_poly r2", test_poly_r2)
```
