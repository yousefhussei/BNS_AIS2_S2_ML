```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([2,2.2,3.6,4.5])
n= len(x)
# initialization
w = 0
b = 0
# print(w, b)
```

```python
##  hyperparameters
learning_rate = 0.01
epochs = 1000
cost_values = []
print(f"initial w:{w}"
      f" initial b:{b}"
      f" learning_rate:{learning_rate}"
      f" epochs: {epochs}")
print("training...................")
```

```python
# Gradient Descent 
# =================
for i in range(epochs):
    #prediction
    y_hat = w * x + b
    #error
    error = y_hat - y
    # gredient 
    D_w = np.mean(error * x)
    D_b = np.mean(error * 1)
    # update weights and bias
    w = w - learning_rate * D_w
    b = b - learning_rate * D_b

    y_hat = w * x + b
    #cost function
    cost = 0.5 * np.mean((y_hat - y) ** 2)
    cost_values.append(cost)

    if (i + 1) % 100 == 0 or i == 0:
        print(f"epoch: {i + 1}"
              f" cost: {cost}"
              f" w: {w}  ,"
              f" b: {b}")

```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(x, w*x+b, color='r')
plt.title("Linear Regression")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.subplot(1,2,2)
plt.plot(cost_values)
plt.title("cost vs Epochs")
plt.xlabel("Epoches")
plt.ylabel("Cost")
plt.grid()
```
